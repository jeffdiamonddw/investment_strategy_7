```python
import pandas as pd
import xarray as xr

df_price = pd.read_parquet('../temp/df_price.parquet')
data = xr.open_dataarray('../temp/data_share_holdings.nc')

sell = - data.diff(dim = 'date').clip(max = 0)
buy = data.diff(dim = 'date').clip(min = 0)


```

```python
df_price_1 = df_price.iloc[:, -sell.shape[0]:].transpose()
price = df_price_1.values.reshape(df_price_1.shape + (1,))
sell_price = sell * price
buy_price = buy * price
cash = (sell_price - buy_price).sum(dim = ['symbol', 'account'])
print(sell_price.sum(dim = ['symbol', 'account']))
print(buy_price.sum(dim = ['symbol', 'account']))
```
